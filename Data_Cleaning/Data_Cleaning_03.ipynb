### Order Reviews Dataset Cleaning

```python
order_reviews_df = pd.read_csv('olist_order_reviews_dataset.csv')
```

```python
order_reviews_df.head()
```

```python
order_reviews_df.shape
```

```python
order_reviews_df.duplicated().sum()
```

##### now because the reviews are not in english we need to translate them.

- we found a ready csv in kaggle that has the english version

```python
reviews_translated_df = pd.read_csv('reviews_translated.csv')
```

```python
reviews_translated_df.head()
```

```python
#ensuring they are the same size
len(order_reviews_df), len(reviews_translated_df)
```

```python
#we will only choose the columns that we need
translated_01 = reviews_translated_df[['review_id', 'order_id', 'review_comment_title', 'review_comment_message']]
```

```python
#we'll change the name of the translated versions so they dont clash with the original columns
translated_01 = translated_01.rename(columns ={
    'review_comment_title' : 'review_comment_title_english',
    'review_comment_message' : 'review_comment_message_english'
})
```

```python
#now we merge
cleaned_order_reviews_dataset = pd.merge(order_reviews_df, translated_01, left_on=['review_id', 'order_id'], right_on=['review_id', 'order_id'], how='left')
```

```python
cleaned_order_reviews_dataset.head()
```

```python
#order the columns in a better way
cleaned_order_reviews_dataset = cleaned_order_reviews_dataset[['review_id', 'order_id', 'review_score', 'review_comment_title', 'review_comment_title_english', 'review_comment_message', 'review_comment_message_english', 'review_creation_date', 'review_answer_timestamp']]
```

```python
cleaned_order_reviews_dataset.head()
```

```python
cleaned_order_reviews_dataset.to_csv('olist_order_reviews_dataset_cleaned.csv', index=False) 
```
