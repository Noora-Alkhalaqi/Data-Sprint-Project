```python
# add new column : name of states:
geolocation["state_name"] = geolocation["geolocation_state"].map(state_names)
```

```python
geolocation["state_name"].value_counts()
```

```python
# change the wrong city and state if the location not in Brazil :
```

```python
# take the maximum and minimum for lat and lng for Brazil :
brazil = (geolocation["geolocation_lat"].between(-33.75, 5.27)& geolocation["geolocation_lng"].between(-73.99, -34.79))

#replace brazil city or state name with international
geolocation.loc[brazil == False, ["geolocation_city","geolocation_state", "state_name"]]="international"
```

```python
#updated state name column:
geolocation["state_name"].value_counts()
```

```python
# check dublicated : 
geolocation.duplicated().sum()
```

```python
geolocation["geolocation_zip_code_prefix"].duplicated().sum()
```

```python
# so, we drop the dublicated 
geolocation=geolocation.drop_duplicates()
```

```python
# check again : 
geolocation.duplicated().sum()
```

```python
# check null value: 
geolocation.isnull().sum()
```

```python
# If there ia a customers not in Brazil :
# ckeck the longitude , latitude from geolocation dataset , then change the wrong state , then replace it in this data 
code = geolocation[geolocation["state_name"] == "international"]["geolocation_zip_code_prefix"]
customers.loc[customers["customer_zip_code_prefix"].isin(code), ["customer_city", "customer_state", "state_name"]] = "international"
```

```python
# check :
customers["state_name"].value_counts()
```

```python
customers.to_csv("olist_geolocation_dataset_cleaned.csv", index = False)
```
